```python
flower_dict = {
    "roses" : list(data_dir.glob("roses/")),
    "daisy" : list(data_dir.glob("daisy/")),
    "dandelion" : list(data_dir.glob("dandelion/")),
    "sunflowers" : list(data_dir.glob("sunflowers/")),
    "tulips" : list(data_dir.glob("tulips/")),

}
```

```python
flower_label = {
    "roses" : 0,
    "daisy" : 1,
    "dandelion" :2,
    "sunflower" : 3,
    "tulips" : 4

}
```

```python
no_of_classes = len(flower_dict.items())
no_of_classes
```

```python
x, y = [], []

for key, value in flower_dict.items():
    print(f"Processing {key} with {len(value)} images")
    for item in value:
        # Skip non-image files
        if item.suffix.lower() not in ['.jpg', '.jpeg', '.png', '.bmp']:
            continue
            
        img = cv2.imread(str(item))
        if img is not None:
            reduced_img = cv2.resize(img, (224, 224))
            x.append(reduced_img)
            y.append(flower_label[key])
        else:
            print(f"[❌] Failed to read image: {item}")

# print(f"Total images processed: {len(x)}")
# print(f"Total labels: {len(y)}")
```

```python
x = np.array(x)
y = np.array(y)
x.shape , y.shape
```

```python
if len(x) > 0:
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
    print(f"Training set: {x_train.shape}, {y_train.shape}")
    print(f"Test set: {x_test.shape}, {y_test.shape}")
else:
    print("No data to split!")
```

```python
x_train_scaled = x_train / 255.0
x_test_scaled = x_test / 255.0
```

```python
##cnn architecture
##conv->maxpool->conv->maxpool->flatten->dense->dense
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical
import numpy as np

# Convert to numpy arrays if not already
x_train = np.array(x_train) / 255.0
x_test = np.array(x_test) / 255.0

# One-hot encode the labels
num_classes = len(np.unique(y_train))
y_train = to_categorical(y_train, num_classes)
y_test = to_categorical(y_test, num_classes)

# CNN model
model = Sequential()

# 1st Conv + MaxPool
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(224, 224, 3)))
model.add(MaxPooling2D(pool_size=(2, 2)))

# 2nd Conv + MaxPool
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Flatten
model.add(Flatten())

# Fully connected layers
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))  # Optional, helps prevent overfitting
model.add(Dense(num_classes, activation='softmax'))  # Output layer

# Compile model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Summary
model.summary()
```

```python
history = model.fit(x_train, y_train, 
                    validation_data=(x_test, y_test),
                    epochs=10, 
                    batch_size=32)
```
