```python
##stemming -- work od this is to reduce the work to its root word
##plays playing played --- play
## runs , running, run --run
```

```python
words = ["plays", "playing" , "played","run","runs" , "running","happened" , "happening" , "happen","History"]
```

```python
## porterstemmer this class helps in this
## the output word may bot be always be  valid english word
from nltk.stem import PorterStemmer
ps = PorterStemmer()

for word in words:
    print(word,"----->",ps.stem(word))
```

```python
## regex stemmer -- regex is a pattern matching language
from nltk.stem import RegexpStemmer
rs = RegexpStemmer("ing$|s$|e$|able$|ibles$")
for word in words:
    print(word,"--------->",rs.stem(word))
```

```python
## snowballstemmer ------> 
from nltk.stem import SnowballStemmer
ss = SnowballStemmer("english")

for word in words:
    print(word,"----->",ss.stem(word))
```
