```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required models
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')

corpus = "Bhagat Singh was a revolutionary freedom fighter. He fought against British rule."

# Sentence tokenization
sent = sent_tokenize(corpus)

# Stemmer and stopword list
stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

# Process each sentence
for i in range(len(sent)):
    words = word_tokenize(sent[i])
    words = [stemmer.stem(word) for word in words if word.lower() not in stop_words]
    pos_tag = nltk.pos_tag(words)  # POS tagging
    print(pos_tag)
```
